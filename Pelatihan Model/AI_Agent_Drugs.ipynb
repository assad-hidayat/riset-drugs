```python
import pandas as pd

# Load dataset
file_path = "data_obat.xlsx"
df = pd.read_excel(file_path)
df
```

```python
import re

# Isi NaN dengan string kosong
for col in df.columns:
    df[col] = df[col].fillna("")

# Fungsi cleaning

def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"\\n", " ", text)
    text = re.sub(r"[^a-zA-Z0-9À-ÿ ]", " ", text)
    text = re.sub(r"\\s+", " ", text).strip()
    return text

# Gabungkan semua informasi menjadi satu dokumen

df["combined_text"] = df.apply(lambda row: f'''
Nama Obat: {row['name_master_erp']}
Kategori: {row['Golongan Obat']}
Kandungan: {row['Bahan Aktif/Kandungan/Komposisi']}
Indikasi: {row['Indikasi']}
Dosis: {row['Dosis Penggunaan']}
Penyajian: {row['Penyajian']}
Kontraindikasi: {row['Kontraindikasi']}
Efek Samping: {row['Efek Samping']}
Cara Kerja: {row['Cara Kerja Obat']}
Harga: {row['Harga']}
''', axis=1)

# Cleaning

df["combined_text"] = df["combined_text"].apply(clean_text)

df[["name_master_erp", "combined_text"]].head()
```

```python
from sentence_transformers import SentenceTransformer

# Load embedding model
embedding_model = SentenceTransformer(
    "paraphrase-multilingual-MiniLM-L12-v2"
)

# Membuat embedding
texts = df["combined_text"].tolist()
embeddings = embedding_model.encode(
    texts,
    show_progress_bar=True,
    convert_to_numpy=True
)

print("Shape embedding:", embeddings.shape)
```

```python
import faiss
import numpy as np

# Dimensi embedding
embedding_dimension = embeddings.shape[1]

# Membuat index FAISS
index = faiss.IndexFlatL2(embedding_dimension)

# Tambahkan embedding ke index
index.add(np.array(embeddings, dtype=np.float32))

print("Jumlah vector dalam index:", index.ntotal)
```

```python
# Simpan index
faiss.write_index(index, "faiss_index.index")

# Simpan dataframe

df.to_pickle("data_obat.pkl")

print("FAISS index berhasil disimpan")
```

```python
import faiss
import pandas as pd

# Load index
index = faiss.read_index("faiss_index.index")

# Load dataframe

df = pd.read_pickle("data_obat.pkl")
```

```python
df
```

```python
import numpy as np

# Fungsi retrieval

def retrieve_documents(query, top_k=3):
    # Embedding query
    query_embedding = embedding_model.encode(
        [query],
        convert_to_numpy=True
    )

    # Search ke FAISS
    distances, indices = index.search(
        np.array(query_embedding, dtype=np.float32),
        top_k
    )

    results = []

    for idx in indices[0]:
        results.append({
            "nama_obat": df.iloc[idx]["name_master_erp"],
            "document": df.iloc[idx]["combined_text"]
        })

    return results
```

```python
query = "pencegah kehamilan"

results = retrieve_documents(query)

for i, res in enumerate(results):
    print(f"\\n=== HASIL {i+1} ===")
    print("Nama Obat:", res["nama_obat"])
    print(res["document"][:500])
```

# Jika Menggunakan Llama

```python
from transformers import pipeline

# Load local LLM
llm = pipeline(
    "text-generation",
    model="microsoft/Phi-3-mini-4k-instruct",
    device_map="auto"
)


def ask_local_llm(query):
    retrieved_docs = retrieve_documents(query, top_k=3)

    context = "\\n\\n".join([
        doc["document"] for doc in retrieved_docs
    ])

    prompt = f'''
Kamu adalah AI Agent Farmasi.

Berikut data obat:
{context}

Pertanyaan:
{query}

Jawaban:
'''

    result = llm(
        prompt,
        max_new_tokens=300,
        do_sample=True,
        temperature=0.3
    )

    return result[0]["generated_text"]
```

```python
question = "Apa obat yang cocok untuk luka bakar ringan?"

answer = ask_local_llm(question)

print(answer)
```
